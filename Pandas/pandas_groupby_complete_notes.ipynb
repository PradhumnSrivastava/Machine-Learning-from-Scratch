
# **Pandas GroupBy**

`groupby()` is one of the most powerful functions in pandas.  
It is used to **split the data into groups**, perform calculations on each group, and combine the results.

This concept is often described as:

**Split → Apply → Combine**

Steps:

1. Split the dataset into groups
2. Apply aggregation functions
3. Combine results


## **1. Import Libraries**

```python

import pandas as pd
import numpy as np
```

## **2. Create Sample Dataset**

```python

data = {
    "Department":["IT","IT","HR","HR","Finance","Finance"],
    "Employee":["A","B","C","D","E","F"],
    "Salary":[50000,60000,45000,55000,70000,65000],
    "Experience":[2,3,5,4,6,7]
}

df = pd.DataFrame(data)

df
```

## **3. Basic GroupBy**

```python

df.groupby("Department")
```


This groups the dataset based on the **Department** column.


## **4. GroupBy with Mean**

```python

df.groupby("Department").mean(numeric_only=True)
```

## **5. GroupBy with Sum**

```python

df.groupby("Department").sum(numeric_only=True)
```

## **6. GroupBy with Count**

```python

df.groupby("Department").count()
```

## **7. GroupBy with Multiple Aggregations**

```python

df.groupby("Department").agg({
    "Salary":["mean","max","min"],
    "Experience":"mean"
})
```

## **8. GroupBy Multiple Columns**

```python

df.groupby(["Department","Employee"]).sum(numeric_only=True)
```

## **9. Selecting Specific Column After GroupBy**

```python

df.groupby("Department")["Salary"].mean()
```

## **10. Iterating Through Groups**

```python

for name, group in df.groupby("Department"):
    print("Department:", name)
    print(group)
```

## **11. GroupBy with Apply Function**

```python

df.groupby("Department")["Salary"].apply(lambda x: x.max() - x.min())
```

## **12. GroupBy with Transform**

```python

df["AverageSalary"] = df.groupby("Department")["Salary"].transform("mean")

df
```

## **13. GroupBy with Size**

```python

df.groupby("Department").size()
```

## **14. Reset Index After GroupBy**

```python

df.groupby("Department").mean(numeric_only=True).reset_index()
```


# **Summary of Important GroupBy Operations**

Important operations:

- groupby()
- mean()
- sum()
- count()
- agg()
- transform()
- apply()
- size()
- grouping multiple columns

GroupBy is widely used for **data aggregation and analysis** in data science.
