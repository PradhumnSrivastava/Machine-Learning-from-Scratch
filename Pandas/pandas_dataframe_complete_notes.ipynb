
# **Pandas DataFrame**

A **Pandas DataFrame** is a two‑dimensional labeled data structure used for handling tabular data.

It is the most commonly used data structure in data science.

Structure of DataFrame:

Rows → index  
Columns → column names  
Values → actual data

Example:

| Name | Age | Salary |
|-----|-----|------|
| A | 25 | 50000 |
| B | 30 | 60000 |


## **1. Import Required Libraries**

```python

import pandas as pd
import numpy as np
```

## **2. Create DataFrame from Dictionary**

```python

data = {
    "Name":["A","B","C","D"],
    "Age":[25,30,35,40],
    "Salary":[50000,60000,70000,80000]
}

df = pd.DataFrame(data)

df
```

## **3. Create DataFrame from List of Lists**

```python

data = [
    ["A",25,50000],
    ["B",30,60000],
    ["C",35,70000]
]

df = pd.DataFrame(data, columns=["Name","Age","Salary"])

df
```

## **4. View First Rows**

```python

df.head()
```

## **5. View Last Rows**

```python

df.tail()
```

## **6. DataFrame Information**

```python

df.info()
```

## **7. Summary Statistics**

```python

df.describe()
```

## **8. Accessing Columns**

```python

df["Name"]
```

## **9. Access Multiple Columns**

```python

df[["Name","Salary"]]
```

## **10. Access Rows using loc**

```python

df.loc[0]
```

## **11. Access Rows using iloc**

```python

df.iloc[0]
```

## **12. Filtering Data**

```python

df[df["Age"] > 30]
```

## **13. Adding New Column**

```python

df["Bonus"] = df["Salary"] * 0.1

df
```

## **14. Dropping Columns**

```python

df.drop("Bonus", axis=1)
```

## **15. Sorting Data**

```python

df.sort_values(by="Age")
```

## **16. Handling Missing Values**

```python

df = pd.DataFrame({
    "A":[1,2,np.nan],
    "B":[4,np.nan,6]
})

df.isnull()
df.dropna()
df.fillna(0)
```

## **17. GroupBy Operation**

```python

data = {
    "Department":["IT","IT","HR","HR"],
    "Salary":[50000,60000,45000,55000]
}

df = pd.DataFrame(data)

df.groupby("Department").mean()
```

## **18. Merge DataFrames**

```python

df1 = pd.DataFrame({
    "ID":[1,2,3],
    "Name":["A","B","C"]
})

df2 = pd.DataFrame({
    "ID":[1,2,3],
    "Salary":[50000,60000,70000]
})

pd.merge(df1, df2, on="ID")
```

## **19. Value Counts**

```python

df = pd.DataFrame({
    "Department":["IT","HR","IT","Finance"]
})

df["Department"].value_counts()
```

## **20. Apply Function on DataFrame**

```python

df = pd.DataFrame({
    "Numbers":[1,2,3,4]
})

df["Square"] = df["Numbers"].apply(lambda x: x**2)

df
```
