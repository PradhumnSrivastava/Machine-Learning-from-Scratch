
# **Pandas Merging and Joining — Complete Essential Codes**

In data analysis, data is often stored in multiple tables.  
Pandas provides powerful functions to **combine datasets**.

Main methods:

1. **merge()** → database-style joins  
2. **join()** → join on index  
3. **concat()** → combine DataFrames vertically or horizontally


## **1. Import Libraries**

```

import pandas as pd
import numpy as np
```

## **2. Create Sample DataFrames**

```

df1 = pd.DataFrame({
    "ID":[1,2,3,4],
    "Name":["A","B","C","D"]
})

df2 = pd.DataFrame({
    "ID":[1,2,3,4],
    "Salary":[50000,60000,70000,80000]
})

df1, df2
```

## **3. Basic Merge**

```

pd.merge(df1, df2, on="ID")
```


This performs an **inner join** by default.


## **4. Inner Join**

```

pd.merge(df1, df2, on="ID", how="inner")
```

## **5. Left Join**

```

pd.merge(df1, df2, on="ID", how="left")
```

## **6. Right Join**

```

pd.merge(df1, df2, on="ID", how="right")
```

## **7. Outer Join**

```

pd.merge(df1, df2, on="ID", how="outer")
```

## **8. Merge on Different Column Names**

```

df3 = pd.DataFrame({
    "EmpID":[1,2,3],
    "Department":["IT","HR","Finance"]
})

pd.merge(df1, df3, left_on="ID", right_on="EmpID")
```

## **9. Merge Multiple Columns**

```

df4 = pd.DataFrame({
    "ID":[1,2,3],
    "Name":["A","B","C"],
    "Age":[25,30,35]
})

pd.merge(df1, df4, on=["ID","Name"])
```

## **10. Merge with Indicator**

```

pd.merge(df1, df2, on="ID", how="outer", indicator=True)
```

## **11. Joining DataFrames using join()**

```

df1.set_index("ID").join(df2.set_index("ID"))
```

## **12. Concatenate DataFrames Vertically**

```

pd.concat([df1, df2])
```

## **13. Concatenate DataFrames Horizontally**

```

pd.concat([df1, df2], axis=1)
```

## **14. Concatenate with Keys**

```

pd.concat([df1, df2], keys=["First","Second"])
```


# **Summary of Combining DataFrames**

Important functions:

- **merge()** → SQL style joins  
- **join()** → index-based join  
- **concat()** → stacking DataFrames

Join Types:

- inner
- left
- right
- outer

These operations are widely used in **data preprocessing and data integration tasks**.
