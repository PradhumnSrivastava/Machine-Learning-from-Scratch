
# **Pandas String Operations & Vectorized String Methods**

Pandas provides powerful **vectorized string operations** through the `.str` accessor.
These operations allow us to efficiently manipulate text data in an entire column without writing loops.

Vectorized string operations are important for:

- Text cleaning
- Feature engineering
- Data preprocessing
- NLP tasks


## **1. Import Libraries**

```python

import pandas as pd
import numpy as np
```

## **2. Create Sample Dataset**

```python

data = {
    "Name":["Alice","Bob","Charlie","David"],
    "City":["New York","London","Paris","Tokyo"],
    "Email":[
        "alice@gmail.com",
        "bob@yahoo.com",
        "charlie@hotmail.com",
        "david@gmail.com"
    ]
}

df = pd.DataFrame(data)
df
```

## **3. Convert Strings to Lowercase**

```python

df["Name"].str.lower()
```

## **4. Convert Strings to Uppercase**

```python

df["Name"].str.upper()
```

## **5. Capitalize Strings**

```python

df["Name"].str.capitalize()
```

## **6. Get Length of Strings**

```python

df["Name"].str.len()
```

## **7. Remove Leading and Trailing Spaces**

```python

df["Name"].str.strip()
```

## **8. Replace Substrings**

```python

df["City"].str.replace("New York","NY")
```

## **9. Check if String Contains a Pattern**

```python

df["Email"].str.contains("gmail")
```

## **10. Startswith Method**

```python

df["Email"].str.startswith("alice")
```

## **11. Endswith Method**

```python

df["Email"].str.endswith("gmail.com")
```

## **12. Split Strings**

```python

df["Email"].str.split("@")
```

## **13. Extract Domain from Email**

```python

df["Domain"] = df["Email"].str.split("@").str[1]
df
```

## **14. Concatenate String Columns**

```python

df["Info"] = df["Name"] + " lives in " + df["City"]
df
```

## **15. Extract Pattern using Regular Expressions**

```python

df["Email"].str.extract(r"(\w+)@")
```

## **16. Count Occurrences of Pattern**

```python

df["Email"].str.count("a")
```

## **17. Find Unique Domains**

```python

df["Domain"].unique()
```
