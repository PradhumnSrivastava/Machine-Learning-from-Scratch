
# **Pandas Series**

A **Pandas Series** is a one-dimensional labeled array capable of holding any data type.

It is one of the two core data structures in pandas:

1. **Series (1D)**
2. **DataFrame (2D)**

A Series contains:

- **Values**
- **Index**


## **1. Import Pandas Library**

```python

import pandas as pd
import numpy as np
```

## **2. Creating a Series from List**

```python

data = [10,20,30,40]
s = pd.Series(data)
s
```

## **3. Creating Series with Custom Index**

```python

s = pd.Series([10,20,30], index=["a","b","c"])
s
```

## **4. Creating Series from Dictionary**

```python

data = {"a":10,"b":20,"c":30}
s = pd.Series(data)
s
```

## **5. Accessing Series Elements**

```python

s["a"]
```

## **6. Accessing Multiple Elements**

```python

s[["a","c"]]
```

## **7. Series Attributes**

```python

s.index
s.values
s.shape
s.size
```

## **8. Mathematical Operations**

```python

s = pd.Series([1,2,3,4])
s + 10
```

## **9. Vectorized Operations**

```python

s * 2
```

## **10. Filtering Series Data**

```python

s[s > 2]
```

## **11. Handling Missing Values**

```python

s = pd.Series([1,2,np.nan,4])
s.isnull()
s.dropna()
```

## **12. Statistical Operations**

```python

s = pd.Series([10,20,30,40])
s.mean()
s.sum()
s.max()
s.min()
```

## **13. Sorting Series**

```python

s = pd.Series([3,1,4,2])
s.sort_values()
```

## **14. Unique Values**

```python

s = pd.Series([1,2,2,3,3,3])
s.unique()
```

## **15. Value Counts**

```python

s.value_counts()
```

## **16. Apply Function on Series**

```python

s = pd.Series([1,2,3,4])
s.apply(lambda x: x**2)
```

## **17. Convert Series to DataFrame**

```python

s.to_frame()
```
