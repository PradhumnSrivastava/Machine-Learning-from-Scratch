
# **Pandas Pivot Tables — Complete Notes and Essential Codes**

A **Pivot Table** is used to summarize and analyze data by grouping values based on categories.

It allows you to:

- Aggregate data
- Reshape datasets
- Perform quick data analysis

A pivot table works similar to **Excel Pivot Tables**.

Basic syntax:

`pd.pivot_table(data, values, index, columns, aggfunc)`


## **1. Import Libraries**

```

import pandas as pd
import numpy as np
```

## **2. Create Sample Dataset**

```

data = {
    "Department":["IT","IT","HR","HR","Finance","Finance"],
    "Employee":["A","B","C","D","E","F"],
    "Salary":[50000,60000,45000,55000,70000,65000],
    "Experience":[2,3,5,4,6,7]
}

df = pd.DataFrame(data)
df
```

## **3. Basic Pivot Table**

```

pd.pivot_table(
    df,
    values="Salary",
    index="Department"
)
```


This groups the dataset by **Department** and calculates the mean salary.


## **4. Pivot Table with Multiple Aggregations**

```

pd.pivot_table(
    df,
    values="Salary",
    index="Department",
    aggfunc=["mean","max","min"]
)
```

## **5. Pivot Table with Columns Parameter**

```

pd.pivot_table(
    df,
    values="Salary",
    index="Department",
    columns="Employee"
)
```

## **6. Pivot Table with Multiple Values**

```

pd.pivot_table(
    df,
    values=["Salary","Experience"],
    index="Department"
)
```

## **7. Pivot Table with Fill Values**

```

pd.pivot_table(
    df,
    values="Salary",
    index="Department",
    columns="Employee",
    fill_value=0
)
```

## **8. Pivot Table with Margins (Totals)**

```

pd.pivot_table(
    df,
    values="Salary",
    index="Department",
    margins=True
)
```

## **9. Pivot Table using Different Aggregation**

```

pd.pivot_table(
    df,
    values="Experience",
    index="Department",
    aggfunc="sum"
)
```

## **10. Pivot Table with Multiple Index Columns**

```

pd.pivot_table(
    df,
    values="Salary",
    index=["Department","Employee"]
)
```


# **Difference Between GroupBy and Pivot Table**

GroupBy:
- More flexible
- Used for complex data transformations

Pivot Table:
- Easier for quick data summaries
- Better for tabular analysis



# **Summary of Important Pivot Table Parameters**

| Parameter | Meaning |
|---|---|
| values | column to aggregate |
| index | row grouping |
| columns | column grouping |
| aggfunc | aggregation function |
| fill_value | replace missing values |
| margins | show totals |
