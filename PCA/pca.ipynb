
# **Principal Component Analysis (PCA)**

Principal Component Analysis (PCA) is a **dimensionality reduction technique** used to reduce the number of features in a dataset while preserving as much information as possible.

PCA transforms the original features into a new set of **uncorrelated variables called Principal Components**.

Main goals of PCA:

- Reduce dimensionality
- Remove multicollinearity
- Improve model performance
- Visualize high‑dimensional data



# **1. Intuition Behind PCA**

Suppose we have a dataset with many features.

Some features may contain **redundant information** or may be **highly correlated**.

PCA solves this by:

1. Finding directions where the data varies the most.
2. Projecting the data onto those directions.

These directions are called **Principal Components**.

Key idea:

First Principal Component → Maximum variance direction  
Second Principal Component → Second highest variance and orthogonal to the first



# **2. Mathematical Idea of PCA**

Steps involved in PCA:

1. Standardize the data
2. Compute covariance matrix
3. Compute eigenvalues and eigenvectors
4. Sort eigenvectors by eigenvalues
5. Project data onto principal components



## **Step 1 — Standardize the Dataset**

PCA is sensitive to scale, so we standardize the data.

Formula:

x' = (x − μ) / σ


```python

import numpy as np
import pandas as pd

from sklearn.preprocessing import StandardScaler

data = pd.DataFrame({
    "age":[25,30,35,40,45],
    "salary":[50000,60000,80000,90000,100000],
    "experience":[1,3,5,7,9]
})

scaler = StandardScaler()

scaled_data = scaler.fit_transform(data)

scaled_data
```


## **Step 2 — Compute Covariance Matrix**

The covariance matrix shows how variables vary together.

Formula:

Cov(X,Y) = Σ (Xi − X̄)(Yi − Ȳ) / (n − 1)


```python

cov_matrix = np.cov(scaled_data.T)

cov_matrix
```


## **Step 3 — Compute Eigenvalues and Eigenvectors**

Eigenvectors determine the **direction of principal components**.

Eigenvalues determine the **amount of variance captured**.


```python

eigenvalues, eigenvectors = np.linalg.eig(cov_matrix)

print("Eigenvalues:", eigenvalues)
print("Eigenvectors:", eigenvectors)
```


## **Step 4 — Sort Components by Explained Variance**

Principal components are sorted based on **largest eigenvalues**.


```python

sorted_indices = np.argsort(eigenvalues)[::-1]

sorted_eigenvalues = eigenvalues[sorted_indices]

sorted_eigenvalues
```


# **3. PCA using Scikit‑Learn**


```python

from sklearn.decomposition import PCA

pca = PCA(n_components=2)

principal_components = pca.fit_transform(scaled_data)

principal_components
```


## **Explained Variance Ratio**

Explained variance shows how much information each component retains.


```python

pca.explained_variance_ratio_
```


# **4. Visualizing PCA**


```python

import matplotlib.pyplot as plt

plt.scatter(principal_components[:,0], principal_components[:,1])

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")

plt.title("PCA Visualization")

plt.show()
```


# **5. Choosing Number of Components**


```python

pca = PCA()

pca.fit(scaled_data)

explained_variance = pca.explained_variance_ratio_

explained_variance
```


## **Cumulative Explained Variance**


```python

cumulative_variance = np.cumsum(explained_variance)

cumulative_variance
```


# **6. Scree Plot**


```python

plt.plot(cumulative_variance)

plt.xlabel("Number of Components")
plt.ylabel("Cumulative Explained Variance")

plt.title("Scree Plot")

plt.show()
```


# **7. PCA in Machine Learning Pipeline**


```python

from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("pca", PCA(n_components=2)),
    ("model", LogisticRegression())
])

pipeline.fit(data, [0,1,1,0,1])
```


# **Summary**

Key concepts in PCA:

- Dimensionality reduction
- Eigenvectors define new feature directions
- Eigenvalues measure variance captured
- Explained variance helps choose components

PCA is commonly used for:

- Data compression
- Noise reduction
- Visualization of high dimensional data
