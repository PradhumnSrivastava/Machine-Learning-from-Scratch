
# **Gradient Descent — Essential Codes**

Gradient Descent is an **optimization algorithm** used to minimize a **loss function** by iteratively updating model parameters.

General update rule:

$$
\theta = \theta - \alpha \nabla J(\theta)
$$

Where

- $\theta$ → parameters (weights)
- $\alpha$ → learning rate
- $\nabla J(\theta)$ → gradient of the cost function



## **1. Basic Gradient Descent Example**

Minimize the function

$$
f(x) = x^2
$$

The minimum occurs at **x = 0**.


```python

import numpy as np

def f(x):
    return x**2

def df(x):
    return 2*x

x = 5
learning_rate = 0.1
iterations = 20

for i in range(iterations):
    
    grad = df(x)
    
    x = x - learning_rate * grad
    
    print(f"Iteration {i+1}: x = {x:.4f}, f(x) = {f(x):.4f}")
```


## **2. Gradient Descent for Linear Regression (From Scratch)**

Linear regression model:

$$
\hat{y} = wx + b
$$

Cost Function:

$$
J(w,b) = \frac{1}{n} \sum (y - \hat{y})^2
$$


```python

import numpy as np

X = np.array([1,2,3,4,5])
y = np.array([2,4,6,8,10])

w = 0
b = 0

learning_rate = 0.01
iterations = 1000
n = len(X)

for i in range(iterations):
    
    y_pred = w*X + b
    
    dw = (-2/n) * np.sum(X*(y - y_pred))
    db = (-2/n) * np.sum(y - y_pred)
    
    w = w - learning_rate * dw
    b = b - learning_rate * db

print("Weight:", w)
print("Bias:", b)
```


## **3. Tracking Cost During Training**

Monitoring the cost helps determine whether the model is converging.


```python

costs = []

for i in range(iterations):

    y_pred = w*X + b
    
    cost = np.mean((y - y_pred)**2)
    
    costs.append(cost)
```


## **4. Vectorized Gradient Descent**

Matrix form:

$$
\hat{y} = XW
$$

Gradient:

$$
\nabla J = \frac{2}{n} X^T (XW - y)
$$


```python

X = np.random.rand(100,2)
true_w = np.array([3,5])

y = X @ true_w + np.random.randn(100)*0.5

w = np.zeros(2)

learning_rate = 0.1
iterations = 500
n = len(X)

for i in range(iterations):
    
    y_pred = X @ w
    
    gradient = (2/n) * X.T @ (y_pred - y)
    
    w = w - learning_rate * gradient

print("Weights:", w)
```


## **5. Batch Gradient Descent**

Uses the **entire dataset** for each update step.


```python

def batch_gradient_descent(X, y, lr=0.01, iterations=1000):

    n, m = X.shape
    
    w = np.zeros(m)

    for i in range(iterations):
        
        y_pred = X @ w
        
        gradient = (2/n) * X.T @ (y_pred - y)
        
        w -= lr * gradient

    return w
```


## **6. Stochastic Gradient Descent (SGD)**

Updates parameters **one sample at a time**.


```python

def stochastic_gradient_descent(X, y, lr=0.01, epochs=50):

    n, m = X.shape
    
    w = np.zeros(m)

    for epoch in range(epochs):

        for i in range(n):

            xi = X[i]
            yi = y[i]

            y_pred = np.dot(xi, w)

            gradient = 2 * xi * (y_pred - yi)

            w -= lr * gradient

    return w
```


## **7. Mini-Batch Gradient Descent**

A compromise between **Batch GD** and **SGD**.


```python

def mini_batch_gradient_descent(X, y, lr=0.01, batch_size=16, epochs=50):

    n, m = X.shape
    
    w = np.zeros(m)

    for epoch in range(epochs):

        indices = np.random.permutation(n)
        X = X[indices]
        y = y[indices]

        for i in range(0, n, batch_size):

            X_batch = X[i:i+batch_size]
            y_batch = y[i:i+batch_size]

            y_pred = X_batch @ w

            gradient = (2/len(X_batch)) * X_batch.T @ (y_pred - y_batch)

            w -= lr * gradient

    return w
```


## **8. Gradient Descent with Learning Rate Decay**

Learning rate decreases over time.


```python

initial_lr = 0.1
decay = 0.01

for i in range(iterations):

    lr = initial_lr / (1 + decay*i)

    gradient = (2/n) * X.T @ (X@w - y)

    w -= lr * gradient
```


## **9. Stopping Criteria**

Training stops when gradient becomes very small.


```python

tolerance = 1e-6

for i in range(iterations):

    gradient = (2/n) * X.T @ (X@w - y)

    if np.linalg.norm(gradient) < tolerance:
        break

    w -= learning_rate * gradient
```


## **10. Practical Gradient Descent using Scikit-Learn**


```python

from sklearn.linear_model import SGDRegressor

model = SGDRegressor(
    learning_rate='constant',
    eta0=0.01,
    max_iter=1000
)

model.fit(X, y)

print(model.coef_)
```
