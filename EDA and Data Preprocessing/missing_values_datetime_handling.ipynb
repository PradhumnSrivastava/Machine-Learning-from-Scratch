
# **Handling Missing Values & Date-Time Features**
Real-world datasets often contain:

- Missing values
- Date & time columns
- Categorical variables
- Numerical variables

Proper preprocessing helps machine learning models work correctly.



## **1. Import Required Libraries**


```python

import pandas as pd
import numpy as np

from sklearn.impute import SimpleImputer
from sklearn.impute import KNNImputer
from sklearn.impute import MissingIndicator
```


## **2. Create Sample Dataset**


```python

data = pd.DataFrame({
    "age":[25, np.nan, 35, 40, np.nan],
    "salary":[50000,60000,np.nan,80000,90000],
    "city":["Delhi","Mumbai",np.nan,"Delhi","Chennai"],
    "date":["2024-01-01","2024-02-10","2024-03-15","2024-04-01","2024-05-20"]
})

data
```


# **Handling Missing Numerical Data**



## **3. Mean Imputation**


```python

imputer = SimpleImputer(strategy="mean")

data["age"] = imputer.fit_transform(data[["age"]])

data
```


## **4. Median Imputation**


```python

imputer = SimpleImputer(strategy="median")

data["salary"] = imputer.fit_transform(data[["salary"]])

data
```


## **5. KNN Imputation**
Uses nearest neighbors to fill missing values.


```python

knn_imputer = KNNImputer(n_neighbors=2)

numeric_data = data[["age","salary"]]

data[["age","salary"]] = knn_imputer.fit_transform(numeric_data)

data
```


# **Handling Missing Categorical Data**



## **6. Most Frequent Imputation**


```python

imputer = SimpleImputer(strategy="most_frequent")

data["city"] = imputer.fit_transform(data[["city"]])

data
```


## **7. Fill Missing with Constant**


```python

data["city"] = data["city"].fillna("Unknown")

data
```


# **Missing Value Indicators**
Creates a binary column showing where data was missing.



## **8. Missing Indicator**


```python

indicator = MissingIndicator()

missing_flags = indicator.fit_transform(data[["age","salary"]])

print(missing_flags)
```


# **Date and Time Handling**
Date columns can be converted to useful features.



## **9. Convert to DateTime**


```python

data["date"] = pd.to_datetime(data["date"])

data
```


## **10. Extract Date Features**


```python

data["year"] = data["date"].dt.year
data["month"] = data["date"].dt.month
data["day"] = data["date"].dt.day

data
```


## **11. Extract Weekday**


```python

data["weekday"] = data["date"].dt.day_name()

data
```


## **12. Time Difference Feature**


```python

data["days_since_start"] = (data["date"] - data["date"].min()).dt.days

data
```


# **Summary**

Handling missing values and date features is essential for building robust ML models.

Common techniques:

- Mean / Median imputation
- Most frequent for categorical data
- KNN imputation
- Missing indicators
- Date feature extraction
