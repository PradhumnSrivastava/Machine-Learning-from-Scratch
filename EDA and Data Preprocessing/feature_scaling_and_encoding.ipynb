
# **Feature Scaling and Encoding Techniques — Essential Codes**

Feature scaling and encoding are important preprocessing steps in machine learning.

They help models learn better by:
- Bringing numerical features to a **similar scale**
- Converting **categorical variables into numerical form**



## **1. Standardization (Z‑Score Scaling)**

Standardization transforms data so that:

Mean = 0  
Standard Deviation = 1

Formula:

x' = (x − μ) / σ


```

from sklearn.preprocessing import StandardScaler
import numpy as np

X = np.array([[25,50000],
              [30,60000],
              [35,80000],
              [40,90000]])

scaler = StandardScaler()

X_standardized = scaler.fit_transform(X)

print(X_standardized)
```


## **2. Normalization (Min‑Max Scaling)**

Normalization scales data into a fixed range.

Most common range:

0 → 1

Formula:

x' = (x − min(x)) / (max(x) − min(x))


```

from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

X_normalized = scaler.fit_transform(X)

print(X_normalized)
```


## **3. Robust Scaling**

Robust scaling is useful when the dataset contains **outliers**.

It uses **median and interquartile range (IQR)** instead of mean and standard deviation.


```

from sklearn.preprocessing import RobustScaler

scaler = RobustScaler()

X_robust = scaler.fit_transform(X)

print(X_robust)
```


## **4. Ordinal Encoding**

Ordinal encoding converts **categorical features into ordered integers**.

Example:

Low → 0  
Medium → 1  
High → 2


```

from sklearn.preprocessing import OrdinalEncoder
import numpy as np

data = np.array([
    ['Low'],
    ['Medium'],
    ['High'],
    ['Medium']
])

encoder = OrdinalEncoder()

encoded_data = encoder.fit_transform(data)

print(encoded_data)
```


## **5. One‑Hot Encoding**

One‑Hot Encoding converts categorical variables into **binary columns**.

Example:

Color:

Red  
Blue  
Green


```

from sklearn.preprocessing import OneHotEncoder

data = np.array([
    ['Red'],
    ['Blue'],
    ['Green'],
    ['Red']
])

encoder = OneHotEncoder(sparse=False)

encoded = encoder.fit_transform(data)

print(encoded)
```


## **6. One‑Hot Encoding using Pandas**

Another simple method for one‑hot encoding is using **pandas get_dummies()**.


```

import pandas as pd

df = pd.DataFrame({
    'City':['Delhi','Mumbai','Delhi','Chennai']
})

encoded_df = pd.get_dummies(df)

print(encoded_df)
```


## **7. When to Use Each Technique**

Standardization:
- Linear Regression
- Logistic Regression
- SVM
- Neural Networks

Normalization:
- KNN
- Neural Networks
- Distance based algorithms

Ordinal Encoding:
- Categorical variables with **order**

One‑Hot Encoding:
- Categorical variables with **no order**
