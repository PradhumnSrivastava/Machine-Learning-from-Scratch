
# **Exploratory Data Analysis — Essential Codes**

EDA (Exploratory Data Analysis) helps understand the structure of a dataset.

Three common types:

- **Univariate Analysis** → analysis of one variable  
- **Bivariate Analysis** → analysis of two variables  
- **Multivariate Analysis** → analysis of more than two variables



## **1. Import Libraries and Create Sample Dataset**


```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

data = pd.DataFrame({
    "age":[22,25,30,35,40,28,32,45,50,36],
    "salary":[25000,30000,50000,60000,65000,45000,52000,70000,80000,62000],
    "experience":[1,2,5,7,10,4,6,12,15,8]
})

data.head()
```


# **Univariate Analysis**
Univariate analysis examines **one variable at a time**.

Common methods:

- Summary statistics
- Histogram
- Box plot
- Distribution analysis



## **2. Summary Statistics**


```python

data["age"].describe()
```


## **3. Histogram**
Used to visualize the distribution of a single feature.


```python

plt.hist(data["age"])
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Age Distribution")
plt.show()
```


## **4. Box Plot**
Helps identify **outliers**.


```python

plt.boxplot(data["salary"])
plt.title("Salary Boxplot")
plt.show()
```


# **Bivariate Analysis**
Bivariate analysis studies the **relationship between two variables**.

Common techniques:

- Scatter plot
- Correlation
- Group comparison



## **5. Scatter Plot**
Used to understand relationship between two numerical variables.


```python

plt.scatter(data["age"], data["salary"])
plt.xlabel("Age")
plt.ylabel("Salary")
plt.title("Age vs Salary")
plt.show()
```


## **6. Correlation**
Measures the strength of relationship between two variables.


```python
data[["age","salary"]].corr()
```


# **Multivariate Analysis**
Multivariate analysis studies relationships between **multiple variables simultaneously**.



## **7. Correlation Matrix**


```python
data.corr()
```


## **8. Scatter Matrix**
Visualizes relationships among multiple variables.


```python

from pandas.plotting import scatter_matrix

scatter_matrix(data)
plt.show()
```


## **9. Pairwise Relationship using Multiple Scatter Plots**


```python

plt.scatter(data["age"], data["salary"])
plt.show()

plt.scatter(data["age"], data["experience"])
plt.show()

plt.scatter(data["experience"], data["salary"])
plt.show()
```
