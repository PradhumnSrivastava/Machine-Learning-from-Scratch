
# **Binning, Binarization and Discretization — Essential Codes**

These techniques are used in feature engineering to transform numerical data into more useful forms.

- **Binning** → Convert continuous values into intervals (bins)
- **Binarization** → Convert values into binary (0 or 1)
- **Discretization** → Convert continuous variables into categorical groups



## **1. Import Required Libraries**


```python

import numpy as np
import pandas as pd

from sklearn.preprocessing import Binarizer
from sklearn.preprocessing import KBinsDiscretizer
```


## **2. Create Sample Dataset**


```python

data = pd.DataFrame({
    "age":[18,22,25,30,35,40,45,50,55,60],
    "income":[20000,25000,30000,40000,50000,60000,70000,80000,90000,100000]
})

data
```


# **Binning**
Binning divides continuous data into **interval groups**.

Example:

0–20 → Young  
20–40 → Adult  
40–60 → Senior



## **3. Equal Width Binning using pandas**


```python

bins = [0,20,40,60]

labels = ["Young","Adult","Senior"]

data["age_group"] = pd.cut(data["age"], bins=bins, labels=labels)

data
```


## **4. Equal Frequency Binning**
Each bin contains approximately the **same number of observations**.


```python

data["income_bins"] = pd.qcut(data["income"], q=3)

data
```


# **Binarization**
Binarization converts numeric values into **0 or 1 based on a threshold**.



## **5. Binarization using sklearn**


```python

binarizer = Binarizer(threshold=40)

age_array = data[["age"]]

binary_age = binarizer.fit_transform(age_array)

print(binary_age)
```


# **Discretization**
Discretization converts continuous features into **discrete bins** using machine learning tools.



## **6. Discretization using KBinsDiscretizer**


```python

kbins = KBinsDiscretizer(n_bins=3, encode="ordinal", strategy="uniform")

age_discretized = kbins.fit_transform(data[["age"]])

print(age_discretized)
```


## **7. Discretization with One-Hot Encoding**


```python

kbins = KBinsDiscretizer(n_bins=3, encode="onehot-dense", strategy="uniform")

age_discretized = kbins.fit_transform(data[["age"]])

print(age_discretized)
```


## **8. Different Discretization Strategies**

KBinsDiscretizer supports different strategies:

- **uniform** → equal width bins
- **quantile** → equal number of samples in each bin
- **kmeans** → bins created using clustering


```python

kbins = KBinsDiscretizer(n_bins=3, encode="ordinal", strategy="quantile")

age_discretized = kbins.fit_transform(data[["age"]])

print(age_discretized)
```


## **9. When to Use These Techniques**

**Binning**
- Reduce noise in data
- Convert continuous variables into categories

**Binarization**
- Convert numeric features to binary indicators

**Discretization**
- Helpful for models that work better with categorical inputs
