
# **Feature Transformation — Essential Codes**
Feature transformation is used to change the distribution of variables so that machine learning models can learn patterns more effectively.

Common goals:
- Reduce skewness
- Normalize distribution
- Improve model performance
- Stabilize variance



## **1. Log Transformation**
Used when data is **right-skewed**.

Formula:

x' = log(x)


```

import numpy as np
import pandas as pd

data = pd.Series([1,5,10,50,100])

log_transformed = np.log(data)

print(log_transformed)
```


## **2. Log1p Transformation (Safe Log)**
Handles zero values.

Formula:

x' = log(1 + x)


```

log1p_transformed = np.log1p(data)

print(log1p_transformed)
```


## **3. Square Root Transformation**
Useful for moderately skewed data.

Formula:

x' = sqrt(x)


```

sqrt_transformed = np.sqrt(data)

print(sqrt_transformed)
```


## **4. Reciprocal Transformation**
Useful for highly skewed distributions.

Formula:

x' = 1/x


```

reciprocal_transformed = 1/data

print(reciprocal_transformed)
```


## **5. Power Transformation**
General form:

x' = x^λ


```

square_transformed = data ** 2
cube_transformed = data ** 3

print(square_transformed)
print(cube_transformed)
```


## **6. Box-Cox Transformation**
Used to normalize skewed data.

Condition: data must be positive.


```

from scipy.stats import boxcox

data_array = np.array([1,5,10,20,50])

transformed_data, lambda_value = boxcox(data_array)

print(transformed_data)
print("Lambda:", lambda_value)
```


## **7. Yeo-Johnson Transformation**
Works with **negative values and zero**.


```

from sklearn.preprocessing import PowerTransformer

pt = PowerTransformer(method='yeo-johnson')

data_reshaped = data_array.reshape(-1,1)

yeo_transformed = pt.fit_transform(data_reshaped)

print(yeo_transformed)
```


## **8. Quantile Transformation**
Transforms features to a uniform or normal distribution.


```

from sklearn.preprocessing import QuantileTransformer

qt = QuantileTransformer(output_distribution='normal')

quantile_transformed = qt.fit_transform(data_reshaped)

print(quantile_transformed)
```


## **9. Rank Transformation**
Replaces values with their rank order.


```

rank_transformed = data.rank()

print(rank_transformed)
```


## **10. Binarization Transformation**
Converts numerical features into binary values.


```

from sklearn.preprocessing import Binarizer

binarizer = Binarizer(threshold=10)

binary_data = binarizer.fit_transform(data_array.reshape(-1,1))

print(binary_data)
```
