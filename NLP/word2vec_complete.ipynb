
# **Word2Vec**

Word2Vec is a **word embedding technique** used in Natural Language Processing (NLP) to represent words as **dense numerical vectors**.

Unlike Bag of Words or TF‑IDF, Word2Vec captures **semantic meaning and relationships between words**.

Example relationships Word2Vec can learn:

king − man + woman ≈ queen

Words with similar meanings will have **similar vector representations**.



# **1. Why Word Embeddings Are Needed**

Traditional methods like:

- Bag of Words
- TF‑IDF

have limitations:

- They create **very large sparse matrices**
- They do not capture **semantic relationships**
- They treat words as **independent tokens**

Word2Vec solves this by learning **dense vector representations** where:

- Similar words are close in vector space
- Contextual meaning is captured



# **2. Word2Vec Architecture**

Word2Vec has two main architectures:

### **1. CBOW (Continuous Bag of Words)**

Predicts a word using its surrounding context.

Example:

Context: "I love ___ learning"

Model predicts:

"machine"

### **2. Skip‑Gram**

Predicts surrounding context words from a target word.

Example:

Target: "machine"

Predicts:

"I", "love", "learning"



# **3. Import Required Libraries**


```python

import numpy as np
import pandas as pd

from gensim.models import Word2Vec
from nltk.tokenize import word_tokenize
import nltk
```


# **4. Download Tokenization Resources**


```python

nltk.download("punkt")
```


# **5. Create Sample Text Dataset**


```python

sentences = [
    "I love machine learning",
    "machine learning is powerful",
    "I love natural language processing",
    "deep learning is amazing",
    "NLP is a part of artificial intelligence"
]
```


# **6. Tokenize Sentences**

Word2Vec requires input as **list of tokenized sentences**.


```python

tokenized_sentences = [word_tokenize(sentence.lower()) for sentence in sentences]

tokenized_sentences
```


# **7. Train Word2Vec Model**


```python

model = Word2Vec(
    sentences=tokenized_sentences,
    vector_size=100,
    window=5,
    min_count=1,
    workers=4
)

model
```


# **8. Word Vector Representation**


```python

model.wv["learning"]
```


Each word is represented by a **vector of numbers**.

These vectors capture **semantic meaning and relationships** between words.



# **9. Find Similar Words**


```python

model.wv.most_similar("learning")
```


Word2Vec finds words with **similar contexts**.



# **10. Word Similarity Score**


```python

model.wv.similarity("machine", "learning")
```


Similarity values closer to **1** indicate stronger semantic similarity.



# **11. Vocabulary Learned by Word2Vec**


```python

model.wv.index_to_key
```


# **12. Example of Word Arithmetic**


```python

model.wv.most_similar(positive=["learning", "machine"])
```


# **How Word2Vec Captures Meaning**

Word2Vec learns meaning by observing **context words**.

Example:

Sentence 1: "I love machine learning"  
Sentence 2: "machine learning is powerful"

Because "machine" and "learning" appear together often, their vectors become **closer in vector space**.

Words appearing in similar contexts develop **similar embeddings**.



# **Advantages of Word2Vec**

- Captures semantic relationships
- Dense vector representation
- Reduces dimensionality
- Useful for many NLP tasks



# **Limitations of Word2Vec**

- Requires large datasets for best performance
- Cannot handle out‑of‑vocabulary words well
- Static embeddings (same vector for every context)



# **Summary**

Word2Vec Pipeline:

Raw Text  
↓

↓
