
# **Part of Speech (POS) Tagging in NLP**

Part‑of‑Speech (POS) tagging is a technique in Natural Language Processing used to identify the **grammatical role of each word in a sentence**.

Each word is assigned a tag such as:

- **NN** → Noun  
- **VB** → Verb  
- **JJ** → Adjective  
- **RB** → Adverb  
- **DT** → Determiner  

Example:

Sentence:

"I love machine learning"

POS Tags:

I → PRP (Pronoun)  
love → VBP (Verb)  
machine → NN (Noun)  
learning → NN (Noun)

POS tagging helps NLP systems understand **sentence structure and grammar**.



# **Why POS Tagging is Important**

POS tagging is used in many NLP tasks:

- Text parsing
- Named Entity Recognition
- Machine Translation
- Information Extraction
- Question Answering systems

It helps models understand **how words function inside a sentence**.


## **1. Import Required Libraries**

```python

import nltk
from nltk.tokenize import word_tokenize
from nltk.tag import pos_tag
```


These modules are used for:

- **word_tokenize** → splitting sentences into words  
- **pos_tag** → assigning part‑of‑speech tags


## **2. Download Required NLTK Resources**

```python

nltk.download("punkt")
nltk.download("averaged_perceptron_tagger")
```


These resources allow NLTK to perform:

- tokenization
- POS tagging using a pretrained model


## **3. Create Example Sentence**

```python

sentence = "I love studying machine learning and natural language processing"
print(sentence)
```

## **4. Tokenization**

```python

tokens = word_tokenize(sentence)
tokens
```


Tokenization splits the sentence into individual words.


## **5. Apply POS Tagging**

```python

pos_tags = pos_tag(tokens)

pos_tags
```


Each tuple contains:

(word, POS_tag)



# **Common POS Tags**

| Tag | Meaning |
|----|----|
| NN | Noun |
| NNS | Plural noun |
| VB | Verb |
| VBD | Past tense verb |
| JJ | Adjective |
| RB | Adverb |
| PRP | Pronoun |
| DT | Determiner |


## **6. POS Tagging for Multiple Sentences**

```python

sentences = [
    "Machine learning is powerful",
    "NLP helps computers understand language",
    "Deep learning models require large datasets"
]

for sentence in sentences:
    
    tokens = word_tokenize(sentence)
    tags = pos_tag(tokens)
    
    print(tags)
```


# **7. Extract Specific POS Tags**

Sometimes we want only specific word types such as:

- only nouns
- only verbs
- only adjectives


```python

nouns = [word for word, tag in pos_tags if tag.startswith("NN")]

print("Nouns:", nouns)
```


Example above extracts **nouns from a sentence**.



# **8. Using POS Tags for Feature Extraction**


```python

verbs = [word for word, tag in pos_tags if tag.startswith("VB")]

print("Verbs:", verbs)
```


# **Advantages of POS Tagging**

- Helps understand sentence structure
- Improves downstream NLP tasks
- Useful for linguistic analysis



# **Limitations**

- Accuracy depends on language model
- Ambiguous words may receive incorrect tags
- Requires context for better interpretation
