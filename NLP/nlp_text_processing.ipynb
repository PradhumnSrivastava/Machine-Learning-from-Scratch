
# **NLP Text Processing**

Text processing (also called **text preprocessing**) is the first step in most Natural Language Processing (NLP) tasks.

Raw text is messy and contains:

- Uppercase / lowercase differences
- Punctuation
- Numbers
- Stopwords
- Different word forms

Before applying machine learning models, we convert text into a **clean and standardized format**.

Typical NLP text preprocessing pipeline:

Raw Text  
↓  
Lowercasing  
↓  
Removing punctuation / numbers  
↓  
Tokenization  
↓  
Stopword removal  
↓  
Stemming / Lemmatization  
↓  
Clean text ready for vectorization


## **1. Import Required Libraries**

```python

import pandas as pd
import numpy as np
import re
import string

import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer
```


Explanation:

- **re** → used for text cleaning using regular expressions  
- **string** → helps remove punctuation  
- **nltk.tokenize** → used for splitting text into words  
- **stopwords** → removes common words like *the, is, and*  
- **PorterStemmer** → reduces words to root form  
- **WordNetLemmatizer** → converts words into dictionary form


## **2. Download Required NLTK Resources**

```python

nltk.download("punkt")
nltk.download("stopwords")
nltk.download("wordnet")
```


These datasets are required for:

- Tokenization
- Stopword removal
- Lemmatization


## **3. Create Sample Text Data**

```python

text = "I LOVE Machine Learning!!! NLP is amazing, isn't it?"
print(text)
```

## **4. Convert Text to Lowercase**

```python

text_lower = text.lower()
print(text_lower)
```


Why lowercase?

Machine learning models treat **Love** and **love** as different words.
Lowercasing helps normalize the text.


## **5. Remove Punctuation**

```python

text_no_punct = text_lower.translate(str.maketrans('', '', string.punctuation))
print(text_no_punct)
```


Punctuation like:

! , . ? '

usually does not carry useful information for many NLP tasks.


## **6. Remove Numbers using Regular Expressions**

```python

text_no_numbers = re.sub(r"\d+", "", text_no_punct)
print(text_no_numbers)
```


Regular expressions allow us to remove:

- numbers
- special characters
- unwanted patterns


## **7. Tokenization**

```python

tokens = word_tokenize(text_no_numbers)
print(tokens)
```


Tokenization splits text into **individual words (tokens)**.

Example:

"I love machine learning"

becomes:

["I", "love", "machine", "learning"]


## **8. Remove Stopwords**

```python

stop_words = set(stopwords.words("english"))

filtered_words = [word for word in tokens if word not in stop_words]

print(filtered_words)
```


Stopwords are very common words such as:

- the
- is
- and
- in
- of

They appear frequently but usually **do not add much meaning**.


## **9. Stemming**

```python

stemmer = PorterStemmer()

stemmed_words = [stemmer.stem(word) for word in filtered_words]

print(stemmed_words)
```


Stemming reduces words to their **root form**.

Example:

learning → learn  
playing → play  
studies → studi


## **10. Lemmatization**

```python

lemmatizer = WordNetLemmatizer()

lemmatized_words = [lemmatizer.lemmatize(word) for word in filtered_words]

print(lemmatized_words)
```


Lemmatization converts words into their **dictionary form**.

Example:

better → good  
running → run

Lemmatization is usually **more accurate than stemming**.



# **Final Clean Text Pipeline Example**


```python

def preprocess_text(text):
    
    text = text.lower()
    
    text = text.translate(str.maketrans('', '', string.punctuation))
    
    text = re.sub(r"\d+", "", text)
    
    tokens = word_tokenize(text)
    
    stop_words = set(stopwords.words("english"))
    tokens = [word for word in tokens if word not in stop_words]
    
    lemmatizer = WordNetLemmatizer()
    tokens = [lemmatizer.lemmatize(word) for word in tokens]
    
    return tokens


clean_tokens = preprocess_text("I LOVE Machine Learning!!! NLP is amazing, isn't it?")

print(clean_tokens)
```


# **Summary of NLP Text Processing Steps**

1. Lowercasing  
2. Removing punctuation  
3. Removing numbers  
4. Tokenization  
5. Stopword removal  
6. Stemming / Lemmatization  

These steps prepare raw text so it can be converted into **numerical features** using:

- Bag of Words
- TF-IDF
- Word Embeddings
